```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'Score': [85, 90, 95, 80]
})
df
```

```python
df['Passed'] = df['Score'] >= 90
df
```

```python
df.plot(x='Name', y='Score', kind='bar')
plt.show()
```
