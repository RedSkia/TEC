```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
city_pop_file = "https://milliams.com/courses/data_analysis_python/city_pop.csv"

census = pd.read_csv(
    city_pop_file,
    skiprows=5,       # skip top text
    sep=";",          # columns separated by semicolon
    na_values="-1",   # treat -1 as missing
    index_col="year"  # set 'year' as index
)
census
```

```python
census.plot()
plt.show()
```

```python
meantemp_file = "https://milliams.com/courses/data_analysis_python/meantemp_monthly_totals.txt"

meantemp = pd.read_csv(
    meantemp_file,
    skiprows=4,        # skip header lines
    sep="\s+",         # split on any whitespace
    index_col=0        # make the first column the index
)

meantemp


```
