```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()  # Apply seaborn style

# Load data
tips = pd.read_csv("https://milliams.com/courses/data_analysis_python/tips.csv")

# Optional: add calculated columns from previous chapter
tips["bill_per_person"] = tips["total_bill"] / tips["size"]
tips["percent_tip"] = tips["tip"] / tips["total_bill"] * 100

tips.head()
```

```python
# Scatter plot: % tip vs bill per person, color by day
sns.relplot(
    data=tips,
    x="bill_per_person",
    y="percent_tip",
    hue="day",
).set(
    xlabel="Bill per person (£)",
    ylabel="Tip (%)"
)
```

```python
# Distribution of bill per person by time of day
sns.displot(
    data=tips,
    x="bill_per_person",
    hue="time",
    kind="kde",       # smooth density plot
    common_norm=False
)
```

```python
# Box plot: total bill by day
sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    order=["Thur", "Fri", "Sat", "Sun"],
    kind="box"
)

# Violin plot: shows distribution + density
sns.catplot(
    data=tips,
    x="day",
    y="bill_per_person",
    order=["Thur", "Fri", "Sat", "Sun"],
    kind="violin"
)

sns.catplot(
    data=tips,
    x="day",
    y="size",
    order=["Thur", "Fri", "Sat", "Sun"],
    kind="violin"
)
```
